```python
import csv
import statistics
```

Questão 4

1. Quantos funcionários existem na empresa? Quais são as informações
que constam no cadastro?

100 funcionários, com as informações 'ID', 'Salario (R$)', 'Idade', 'Escolaridade', 'No de Filhos'

```python
with open('funcionarios.csv', newline='') as csvfile:
    reader = csv.reader(csvfile, delimiter=',')
    cabecalho = next(reader)
    print("Colunas:", cabecalho)
    dados=list(reader)
    print(len(dados))
    
```

2. Calcule média, variância, desvio padrão dos salários dos funcionários.
Interprete o resultado.

variância: 2556439.39 R$²
média: 4855
desvio padrão: 1598.89 R$

A empresa oferta salários bons para a realidade brasileira com média de aproximadamente 3 salários mínimos, em sua maioria, e não apresenta tantas discrepâncias num geral, a disparidade salarial é excessão e não regra.

```python
idades = []
salarios = []

for linha in dados:
    idade = int(linha[2])          
    salario = float(linha[1])       
    
    idades.append(idade)
    salarios.append(salario)
```

```python
media_salario = statistics.mean(salarios)
var_salario = statistics.variance(salarios)
dp_salario = statistics.stdev(salarios)

print(media_salario)
print(f"{var_salario:.2f}")
print(f"{dp_salario:.2f}")
```

3. Que idade tem o funcionário mais jovem? 

20 anos

```python
print(min(idades))
```

4. Qual é a quantidade máxima de filhos dos funcionários? 

4 filhos

```python
n_filhos = []

for linha in dados:
    filhos = int(linha[4])

    n_filhos.append(filhos)

print(max(n_filhos))
```

5. Qual é a quantidade de filhos mais comum?

1 filho

```python
moda_filhos = int(statistics.mode(n_filhos))

print(moda_filhos)
```

```python
import matplotlib as plt

# Organizando os dados fornecidos em uma lista única
erros = [
    8, 11, 8, 12, 14, 13, 11, 14, 14, 5,
    6, 10, 14, 19, 6, 12, 7, 5, 8, 8,
    10, 16, 10, 12, 12, 8, 11, 6, 7, 12,
    7, 10, 14, 5, 12, 7, 9, 12, 11, 9,
    14, 8, 14, 8, 12, 10, 12, 22, 7, 15
]

dias = list(range(1, 51))

media = statistics.mean(erros)
mediana = statistics.median(erros)
moda = statistics.mode(erros)

variancia = statistics.variance(erros)
desvio_padrao = statistics.stdev(erros)

print(f"Média: {media:.2f}")
print(f"Mediana: {mediana}")
print(f"Moda: {moda}")
print(f"Variância: {variancia:.2f}")
print(f"Desvio Padrão: {desvio_padrao:.2f}")

# i. Geração do gráfico
plt.figure(figsize=(10, 5))
plt.plot(dias, erros, marker='o', linestyle='-', color='#1f77b4')
plt.title('Erros de Impressão por Dia')
plt.xlabel('Dias')
plt.ylabel('Número de Erros')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
